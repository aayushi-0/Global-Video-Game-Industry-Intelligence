```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
```

```python
games = pd.read_csv("../data/raw/Video_Games_Sales_as_at_22_Dec_2016.csv")
```

```python
games_clean = games.copy()
```

```python
games_clean.shape
```

```python
games_clean.isnull().sum()
```

```python
games_clean = games_clean.dropna(subset=['Name'])
```

```python
games_clean.shape
```

```python
games_clean['Name'].isnull().sum()
```

```python
games_clean[games_clean['Genre'].isnull()]
```

```python
games_clean['User_Score'].value_counts().head(10)
```

```python
(games_clean['User_Score'] == 'tbd').sum()
```

```python
games_clean['User_Score'] = games_clean['User_Score'].replace('tbd', np.nan)
```

```python
games_clean['User_Score'] = pd.to_numeric(games_clean['User_Score'])
```

```python
games_clean.to_csv("../data/cleaned/video_games_sales_cleaned.csv", index=False)
```

```python
games_clean["Year_of_Release"] = games_clean["Year_of_Release"].astype("Int64")
games_clean["Critic_Count"] = games_clean["Critic_Count"].astype("Int64")
games_clean["User_Count"] = games_clean["User_Count"].astype("Int64")

games_clean.to_csv(
    "../data/cleaned/video_games_sales_cleaned.csv",
    index=False,
    na_rep="NULL"
)
```
