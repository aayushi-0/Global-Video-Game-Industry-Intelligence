```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)

steam = pd.read_csv("../data/raw/merged_data.csv")

steam_clean = steam.copy()
```

```python
steam_clean = steam_clean.dropna(subset=['Title'])
```

```python
steam_clean['Title'].isnull().sum()
```

```python
steam_clean[steam_clean['Original Price'].str.contains(',', na=False)][['Original Price']].drop_duplicates()
```

```python
steam_clean[steam_clean['Discounted Price'].str.contains(',', na=False)][['Discounted Price']].drop_duplicates()
```

```python
steam_clean['Original Price'] = (
    steam_clean['Original Price']
    .replace('Free', '0')
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
)

steam_clean['Original Price'] = pd.to_numeric(steam_clean['Original Price'])
```

```python
steam_clean[['Original Price', 'Discounted Price']].dtypes
```

```python
temp = (
    steam_clean['Discounted Price']
    .replace('Free', '0')
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
)

pd.to_numeric(temp, errors='coerce')[pd.to_numeric(temp, errors='coerce').isna()]
```

```python
steam_clean.loc[
    pd.to_numeric(temp, errors='coerce').isna(),
    'Discounted Price'
].drop_duplicates()
```

```python
steam_clean['Discounted Price'] = pd.to_numeric(
    steam_clean['Discounted Price']
        .replace('Free', '0')
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False),
    errors='raise'
)
```

```python
steam_clean['Discounted Price'].dtype
```

```python
steam_clean['Discount Percentage'] = np.where(
    steam_clean['Original Price'] > 0,
    ((steam_clean['Original Price'] - steam_clean['Discounted Price'])
     / steam_clean['Original Price']) * 100,
    0
)

steam_clean['Discount Percentage'] = steam_clean['Discount Percentage'].round(2)
```

```python
steam_clean[['Original Price',
             'Discounted Price',
             'Discount Percentage']].head(10)
```

```python
steam_clean['Is Free'] = steam_clean['Original Price'] == 0
```

```python
steam_clean['Release Date'].head(20)
```

```python
steam_clean['Release Date'].sample(20, random_state=42)
```

```python
(steam_clean['Release Date'] == 'Coming soon').sum()
```

```python
steam_clean['Release Date'] = pd.to_datetime(
    steam_clean['Release Date'],
    format='%d %b, %Y',
    errors='coerce'
)
```

```python
steam_clean['Release Date'].dtype
```

```python
steam_clean['Release Date'].isnull().sum()
```

```python
steam_clean['Release Year'] = steam_clean['Release Date'].dt.year

steam_clean['Release Month'] = steam_clean['Release Date'].dt.month_name()
```

```python
steam_clean['Release Status'] = np.where(
    steam_clean['Release Date'].isna(),
    'Coming Soon',
    'Released'
)
```

```python
steam.loc[
    pd.to_datetime(
        steam['Release Date'],
        format='%d %b, %Y',
        errors='coerce'
    ).isna(),
    'Release Date'
].value_counts().head(20)
```

```python
steam_clean['Release Month Number'] = steam_clean['Release Date'].dt.month
```

```python
steam_clean['Recent Reviews Number'].dropna().head(10)
```

```python
steam_clean['All Reviews Number'].dropna().head(10)
```

```python
steam_clean['Recent Review %'] = (
    steam_clean['Recent Reviews Number']
    .str.extract(r'(\d+)%')
    .astype(float)
)
```

```python
steam_clean['Recent Review Count'] = (
    steam_clean['Recent Reviews Number']
    .str.extract(r'([\d,]+)\s+user reviews')[0]
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

```python
steam_clean['All Review %'] = (
    steam_clean['All Reviews Number']
    .str.extract(r'(\d+)%')
    .astype(float)
)
```

```python
steam_clean['All Review Count'] = (
    steam_clean['All Reviews Number']
    .str.extract(r'([\d,]+)\s+user reviews')[0]
    .str.replace(',', '', regex=False)
    .astype(float)
)
```
